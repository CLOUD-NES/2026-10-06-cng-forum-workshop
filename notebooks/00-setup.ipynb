## Write, list, read and delete an object

We use [fsspec](https://filesystem-spec.readthedocs.io) (with [s3fs](https://s3fs.readthedocs.io)) to interact with the object store as if it were a file system.

```python
import fsspec

fs = fsspec.filesystem("s3", endpoint_url=ENDPOINT_URL)
```

```python
test_object = f"{USER_PREFIX_URI}/hello.txt"
fs.pipe(test_object, b"Hello, cloud-native world!")
```

```python
fs.ls(USER_PREFIX_URI)
```

```python
fs.cat(test_object)
```

NOTE: the bucket is publicly readable. Every object can also be retrieved with a plain HTTP(S) request, without credentials. We will use this later to browse our STAC catalog.

```python
from utils import https_url

url = https_url(test_object)
url
```

```python
fs.rm(test_object)
```
